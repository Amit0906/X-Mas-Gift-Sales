```python
# importing pandas
import pandas as pd

# importing dataset from Github
df=pd.read_excel("https://github.com/Amit0906/X-Mas-Gift-Sales/raw/refs/heads/main/Organisation%20Dataset.xlsx")
print(df)
```

```python
# Displaying the information regarding the dataset
df.info()
```

```python
# To print the lowest number of Employees
print("The Lowest number of Employees")
print(df['Number of employees'].min())
```

```python
# To print the highest number of Employees
print("The Highest number of Employees")
print(df['Number of employees'].max())
```

```python
# Finding the correlation between Foundation and Number of Employees
df['Founded'].corr(df['Number of employees'])
```

```python
# importing matplotlib
import matplotlib.pyplot as ply

# creating a line chart with Founded as x-axis, Number of Employees as y-axis
ply.plot(df['Founded'],df['Number of employees'],marker='.')
# labeling X-axis
ply.xlabel("Year of Foundation")
# labeling Y-axis
ply.ylabel("Number of Employees")
# Title of the chart
ply.title("Line Graph representing the Correlation between 'Founded' and 'No. of Employees'")
ply.show()
```
